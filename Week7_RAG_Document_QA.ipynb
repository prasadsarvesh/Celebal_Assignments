```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("/content/resume.pdf")
documents = loader.load()

print("Number of pages:", len(documents))
print(documents[0].page_content[:500])
```

## 3. Split the document into chunks

```python
!pip install -q langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

docs = text_splitter.split_documents(documents)

print("Total chunks:", len(docs))
```

## 4. Create embeddings and build the vector database

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vector_db = FAISS.from_documents(docs, embedding_model)

print("✅ Vector Database Created Successfully!")
```

## 5. Ask questions about the resume

```python
query = "What are the skills mentioned in the resume?"

results = vector_db.similarity_search(query, k=2)

print("Question:", query)
print("\nAnswer:\n")

for i, doc in enumerate(results):
    print(f"Result {i+1}:")
    print(doc.page_content)
    print("-" * 50)
```

## 6. Interactive Q&A loop
Type a question about the resume, or `exit` to stop.

```python
while True:
    query = input("Ask a question about the PDF (type 'exit' to quit): ")

    if query.lower() == "exit":
        break

    results = vector_db.similarity_search(query, k=2)

    print("\nAnswer:\n")
    for i, doc in enumerate(results, 1):
        print(f"Result {i}:")
        print(doc.page_content)
        print("-" * 50)
```
